### ***A. Mètode GET***

1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:
    *   Llista de publicacions (/posts)
    *   Llista d’usuaris (/users)
    *   Llista de tasques (/todos)

2. Mostra per pantalla:
    *   La quantitat total de cada recurs.
    *   El codi d'estat de cada petició.

```python
# DEFINICIÓ DE LA FUNCIÓ

def obtenir_dades (url, text):
# Envia una petició GET a la URL indicada per obtenir-ne el contingut.
# El resultat (la resposta del servidor) es desa a la variable resposta
  print (text)
  resposta = requests.get(url)
  print("Hem obtingut",len(resposta.json()),"dades")
  print ("El codi d'estat de la petició és:", resposta.status_code," -", resposta.reason)
  return resposta.json()

# DADES
url_posts = "https://jsonplaceholder.typicode.com/posts"
url_usuaris = "https://jsonplaceholder.typicode.com/users"
url_tasques = "https://jsonplaceholder.typicode.com/todos"


# EXECUCIÓ
llista_publicacions = obtenir_dades (url_posts, "LLISTA DE PUBLICACIONS")
llista_usuaris = obtenir_dades (url_usuaris, "LLISTA D'USUARIS")
llista_tasques = obtenir_dades (url_tasques, "LLISTA DE TASQUES")
```

### ***B. Mètode POST***

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.
    *   Mostra la resposta JSON.
    *   Mostra el codi d'estat.

```python
# DEFINICIÓ DE LA FUNCIÓ

def enviar_dades (url, dades):
  print ("ENVIAMENT DE DADES")
  resposta = requests.post(url, dades)
  print ("La resposta JSON és:", resposta.json())
  print ("El codi d'estat de la petició és:", resposta.status_code," -", resposta.reason)


# DADES
dades = {"userId": 99, "title": "Prova55", "body": "sunt aut facere repellat provident occaecati excepturi optio reprehenderit"}

# EXECUCIÓ
enviar_dades (url_posts, dades)
```

### ***C. Mètode PATCH***

5. Fes una petició PATCH per modificar parcialment una publicació existent.
    *   Mostra la resposta JSON.
    *   Mostra el codi d'estat.

```python
# DEFINICIÓ DE LA FUNCIÓ

def modificar_dades (url, dades):
  print ("MODIFICACIÓ DE DADES")
  resposta = requests.patch(url, json=dades)
  print ("La resposta JSON és:", resposta.json())
  print ("El codi d'estat de la petició és:", resposta.status_code," -", resposta.reason)


# DADES
# Modifiquem el post 1 afegint el seu ID a la URL
url_patch = "https://jsonplaceholder.typicode.com/posts/1"
dades = {"title": "Prova66", "body": "Hola"}

# EXECUCIÓ
modificar_dades (url_patch, dades)
```

### ***D. Mètode DELETE***

6. Fes una petició DELETE sobre una publicació.
    *   Mostra la resposta JSON.
    *   Mostra el codi d'estat.

```python
# DEFINICIÓ DE LA FUNCIÓ

def esborrar_dades (url):
  print ("ELIMINACIÓ DE DADES")
  resposta = requests.delete(url)
  print ("La resposta JSON és:", resposta.json())
  print ("El codi d'estat de la petició és:", resposta.status_code," -", resposta.reason)


# DADES
# Esborrem  el post 1 afegint el seu ID a la URL
url_del = "https://jsonplaceholder.typicode.com/posts/1"

# EXECUCIÓ
esborrar_dades (url_del)
```

*   url per accedir-hi: https://opendata-ajuntament.barcelona.cat/data/api/action/mètode

Aquests **mètodes** són públics i no requereixen cap tipus d'identificació ni token sent alguns exemples:


*   **current_package_list_with_resources**: retorna el llistat de datasets actuals i els seus recursos
*   **package_search**: fa una cerca dins del catàleg de datasets
*   **resource_show**: retorna les metadades del dataset a partir de l'id del dataset i entre elles el resource_id dels recursos
*   **datastore_search**: retorna la informació de cada recurs.

```python
# Base URL de l'API de CKAN d'Open Data BCN
BASE_URL = "https://opendata-ajuntament.barcelona.cat/data/api/action"
```

### ***1. Cerca de datasets mitjançant package_search***

```python
# -----------------------------------------------------------------------------
# 1. Cerca de datasets mitjançant package_search
# -----------------------------------------------------------------------------
query = "aparcaments"
search_url = f"{BASE_URL}/package_search?q={query}"
cataleg = requests.get(search_url)
# L'API retorna un diccionari amb metadades. Els datasets reals estan a ['result']['results']
print("Hem obtingut",len(cataleg.json()['result']['results']),"dades")
print ("El codi d'estat de la petició és:", cataleg.status_code," -", cataleg.reason)

# Agafem el primer resultat trobat
results = cataleg.json()["result"]["results"]

# Convertim a dataframe i mostrem
df_cataleg = pd.DataFrame(results)
mostrar_dataframe(df_cataleg, "LLISTA DE DATASETS")
```
