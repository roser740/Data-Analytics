### ***✔ obtenir_dades()***

Per emmagatzemar la informació de cada candidat utilitzarem un diccionari que contindrà totes les seves dades.

**Tasques:**
1. Demanar les dades al candidat via input i guardar-les al diccionari amb la clau corresponent (nom, telèfon, antiguitat, català, excel i títol)
2. Retornar diccionari amb totes les dades d'un candidat.


```python
def obtenir_dades1():
    #inicialitzem el diccionari
    dades = {}
    #capturem les dades en format string i les guardem al diccionari
    dades['nom'] = input("Introdueix el teu nom i cognoms: ")
    dades['telèfon'] = input("Introdueix el telèfon: ")
    dades['antiguitat'] = input("Introdueix els anys d'experiència a l'empresa: ")
    dades['català'] = input("Introdueix el nivell de coneixement de català: ")
    dades['excel'] = input("Introdueix el nivell de coneixement d'excel: ")
    dades['títol'] = input("Introdueix el tipus de títol acadèmic: ")
    #retornem el diccionari amb les dades del candidat
    return dades
```

```python
def obtenir_dades(llista_inicial, n):
    #inicialitzem el diccionari amb les dades de la llista_inicial
    dades = {}
    #capturem les dades en format string i les guardem al diccionari
    dades['nom'] = llista_inicial[0+6*n]
    dades['telèfon'] = llista_inicial[1+6*n]
    dades['antiguitat'] = llista_inicial[2+6*n]
    dades['català'] = llista_inicial[3+6*n]
    dades['excel'] = llista_inicial[4+6*n]
    dades['títol'] = llista_inicial[5+6*n]
    #retornem el diccionari amb les dades del candidat
    print(dades)
    return dades
```

### ***✔ validar_dades(dades)***

**Paràmetre:** diccionari amb les dades d'un candidat.

**Tasques:**
1. Processar les dades entrades:
    - **Convertir** els camps numèrics a enters (antiguitat, català, excel i títol).
    - **Validar** que els camps numèrics anteriors estiguin dins del rang permès.

2. Si falla algun procés mostra missatge d'error i retorna **False**.

3. Si no hi ha errors retorna **True**.

```python
def validar_dades(dades):
    try:
        # Convertim els camps numèrics a tipus enter.
        dades['antiguitat'] = int(dades['antiguitat'])
        dades['català'] = int(dades['català'])
        dades['excel'] = int(dades['excel'])
        dades['títol'] = int(dades['títol'])

        # Validem els rangs de les dades numèriques. Si no estan dins del rang permès mostrem missatge i retornem False.
        if not (0 <= dades['català'] <= 3):
            print("Error: El coneixement de català ha de ser entre 0 i 3.")
            return False
        if not (0 <= dades['excel'] <= 3):
            print("Error: El domini d'Excel ha de ser entre 0 i 3.")
            return False
        if not (0 <= dades['títol'] <= 3):
            print("Error: El títol acadèmic ha de ser entre 0 i 3.")
            return False
        if dades['antiguitat'] < 0:
            print("Error: Els anys d'experiència no poden ser negatius.")
            return False

    #si la conversió dona error mostrem missatge i retornem False
    except ValueError:
        print("Error: Els camps numèrics (antiguitat, català, excel, títol) han de ser nombres enters.")
        return False

    #si no hi ha cap error retornem True
    return True
```

### ***✔ Calcular_puntuació(dades)***

**Paràmetre:** diccionari amb les dades d'un candidat.

**Tasques:**
1. Calcular la puntuació: català + excel + títol + anys * 1,5
2. Retornar el resultat

```python
def calcular_puntuació(dades):
    #calculem la puntuació
    puntuació = dades['català'] + dades['excel'] + dades['títol'] + dades['antiguitat'] * 1.5
    #retornem el resultat
    return puntuació
```

### ***✔ classificar_candidat(puntuació)***

**Paràmetre:** valor de la puntuació del candidat.

**Tasques:**
1. Classificar el candidat segons barem:
    *   si puntuació >= 12 --> prioritat alta
    *   si puntuació < 8 --> no prioritari
    *   en els altres casos (8<= puntuació <12) --> candidat adequat
2. Retornar resultat

```python
def classificar_candidat(puntuació):
    #classifiquem el candidat en funció del valor de la puntuació
    if puntuació >= 12:
        classificació = "Prioritat alta"
    elif puntuació < 8:
        classificació = "No prioritari"
    else:
        classificació = "Candidat adequat"

    #retornem el resultat de la classificació
    return classificació
```

### ***✔ ordenar_dataframe(df, camp, ordre)***

```python
def ordenar_dataframe(df, camp, ordre):
    #Modifica el dataframe ordenant-lo pel camp indicat i seguint l'ordre indicat (True: ascendent, False: descendent)
    df.sort_values(by=camp, ascending=ordre, inplace=True)
```

## 🧠 **Algoritme principal**



### ***✔ main_seleccio():***

1. Mentre num usuaris <= 10:
    *   Demanar dades a l'usuari via input: `obtenir_dades()`
    *   Validar que les dades estiguin dins els rangs: `validar_dades()`
    *   Si la validació és correcta:
        *   Calcular puntuació: `calcular_puntuacio()`
        *   Classificar el candidat: `classificar_candidat()`
        *   Mostrar la puntuació individual: `mostrar_individual()`
2. Mostrar informe final: `mostrar_informe()`

```python
def main_seleccio():
    #inicialitzem la llista de diccionaris on guardarem les dades de tots els candidats.
    print("\n" + "="*100)
    print("ENTRADA DE DADES")
    print("="*100)
    candidats_registrats = []
    i = 0
    while i < 10:
        #inicialitzem el diccionari on guardarem les dades del candidat que està entrant la informació.
        dades_candidat = {}
        #capturem les dades del candidat i les guardem al diccionari
        dades_candidat = obtenir_dades(inputs_test, i)
        #si la validació de les dades retorna True calculem puntuació, classifiquem candidat, mostrem puntuació i afegim a la llista de candidats.
        if validar_dades(dades_candidat):
          dades_candidat['puntuació'] = calcular_puntuació(dades_candidat)
          dades_candidat['classificació'] = classificar_candidat(dades_candidat['puntuació'])
          mostrar_individual(dades_candidat['puntuació'])
          candidats_registrats.append(dades_candidat)
        i += 1
    #mostrem informe ordenat dels candidats.
    df_candidats = pd.DataFrame(candidats_registrats)
    ordenar_dataframe(df_candidats, 'puntuació', False)
    mostrar_dataframe(df_candidats[['nom', 'telèfon','puntuació']], "INFORME FINAL DE CANDIDATS")
```

Definim dos diccionaris amb les dades associades al codi comportament i al grup d'edat.

```python
MAPA_COMPORTAMENT = {
    0: {"factor": 0.10, "segment": "Navegador passiu", "missatge": "Descobreix les nostres novetats que et podrien agradar!"},
    1: {"factor": 0.25, "segment": "Interès inicial", "missatge": "Encara hi ets a temps! Els teus productes t'esperen a la cistella."},
    2: {"factor": 0.50, "segment": "Abandó a checkout", "missatge": "Completa la teva comanda ara i t'enviem un codi d'enviament gratuït."},
    3: {"factor": 0.75, "segment": "Abandó recent (<24h)", "missatge": "Has oblidat alguna cosa? Finalitza la teva compra abans que s'exhaureixi l'estoc."},
    4: {"factor": 0.90, "segment": "Abandó crític (>24h)", "missatge": "T'oferim un 10% de descompte exclusiu per recuperar la teva cistella!"}
}

MAPA_EDAT = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}
```

## 🧩 **Descomponent el problema: naixement de les funcions**

### ***♻ Re-utilització de funcions:***

### ***✔ obtenir_valors(clau, diccionari)***

**Paràmetres:** clau i diccionari

**Operacions:** Rep una clau i accedeix a l'estructura del diccionari indicat per extreure directament les dades associades.

**Retorn:** dades associades a la clau

```python
def obtenir_valors(clau, diccionari):
  #Retorna les dades associades a la clau indicada dins del diccionari
  return diccionari.get(clau)
```

### ***✔ calcular_risc(factor codi, factor edat)***

**Paràmetres:** factor de codi comportament i factor d'edat

**Operacions:**  Calcular: risc final = factor codi * factor edat

**Retorn:** el valor de risc final arrodonit.

```python
def calcular_risc(factor_codi, factor_edat):
  #Retorna el risc calculat arrodonit a 2 decimals.
  return round(factor_codi * factor_edat , 2)
```

### ***✔ processar_clients(llista inicial)***

**Paràmetre:** llista de diccionaris amb les dades dels clients.

**Passos:**
1. Per cada client de la llista inicial:
    - Obtenir factor, segment i missatge segons codi comportament --> obtenir_dades()
    - Obtenir factor edat -- > obtenir_dades()
    - Calcular risc --> calcular_risc()
    - Guardar aquestes dades a un nou diccionari.
    - Afegir el nou diccionari del client amb les dades actualitzades a la llista final.

**Retorn:** la llista de clients actualitzada


```python
def processar_dades(client):

  #inicialitzem un nou diccionari i hi copiem les dades actuals del client
  dades_client = {}
  dades_client = client
  #obtenim segment, missatge i factor segons codi comportament i els guardem al nou diccionari
  dades_comportament  = obtenir_valors(client['codi_comportament'], MAPA_COMPORTAMENT)
  dades_client ['segment'] = dades_comportament ['segment']
  dades_client ['missatge'] = dades_comportament ['missatge']
  dades_client ['factor_codi'] = dades_comportament ['factor']
  #obtenim factor segons grup d'edat i el guardem al nou diccionari
  dades_client ['factor_edat'] = obtenir_valors(client['grup_edat'], MAPA_EDAT)
  #calculem risc i el guardem al nou diccionari
  dades_client ['risc_final'] = calcular_risc(dades_client ['factor_codi'], dades_client ['factor_edat'] )

  #retornem diccionari final amb les dades actualitzades
  return dades_client
```

```python
def convertir_llista(llista_inicial):
  #Retorna una llista de diccionaris. Cada diccionari conté les dades dels elements de la llista inicial actualitzades
  llista_final = []
  for element in llista_inicial:
    llista_final.append(processar_dades(element))
  df = pd.DataFrame(llista_final)
  return df
```

### ***✔ netejar_símptomes(llista_inicial)***

**Paràmetres:** llista de símptomes en brut

**Operacions:**
1. Per cada element de la llista:
    *   convertir a mínuscules
    *   netejar espais inicials i finals
    *   convertir espais intermitjos a _

**Retorn:** llista actualitzada de símptomes

```python
def netejar_símptomes(llista_inicial):
  #Retorna

  # Inicialitzem la llista
  llista_final = []
  for símptoma in llista_inicial:
    llista_final.append(símptoma.lower().lstrip().rstrip().replace(" ", "_"))
  return llista_final
```

### ***✔ calcular_risc(llista_símptomes)***

**Paràmetres:** llista de símptomes

**Operacions:**
1. Per cada element de la llista:
    *   risc total = risc total + risc símptoma

**Retorn:** risc total del pacient

```python
def calcular_risc (llista_símptomes):
  risc_pacient = 0
  for símptoma in llista_símptomes:
    if símptoma in nivells_risc:
      risc_pacient += nivells_risc[símptoma]
  return risc_pacient
```

### ***✔ processar_dades(pacient)***

**Paràmetres:** string amb les dades d'un pacient

**Operacions:**
1. Inicialitzar diccionari del pacient
2. Separar camps per |
3. Separar símptomes per , i crear llista
4. Assignar valors al diccionari amb les claus: nom, edat i símptomes

**Retorn:** diccionari amb les dades del pacient actualitzades

```python
def processar_dades(pacient):
  # Transforma un string de dades brutes en un diccionari amb les dades normalitzades

  # Inicialitzem el diccionari on recollirem les dades del pacient
  dades_pacients = {}
  # Separem la cadena inicial pel separador | i obtenim una llista d'elements
  dades = pacient.split("|")
  # Creem una clau anomenada 'nom' dins del diccionari dades_pacients i li assignem el valor del primer element de la llista
  dades_pacients['nom'] = dades[0]
  # Creem una clau anomenada 'edat' dins del diccionari dades_pacients i li assignem el valor del segon element de la llista convertit a enter
  dades_pacients['edat'] = int(dades[1])
  dades_pacients['símptomes'] = netejar_símptomes(dades[2].split(","))
  dades_pacients['risc_total'] = calcular_risc(dades_pacients['símptomes'])

  return dades_pacients
```

### ***✔ caràcters_ascii(codi_ini, codi_fin)***

**Paràmetre:** codi ascii inicial i codi ascii final

**Passos:**

1. Des del codi inicial fins al codi final:
    *   caràcter = chr(codi).
    *   afegir caràcter a la llista final.

**Retorn:** la llista de caràcters associats als codis rebuts.

```python
def caràcters_ascii(codi_ini, codi_fin):
  #Retorna una llista de caràcters associats als codis ascii rebuts
  llista_final = []
  for codi in range(codi_ini, codi_fin+1):
    llista_final.append(chr(codi))
  return llista_final
```

### ***✔ definir_llista(majus, minus, num, signes)***

**Paràmetres:** valors booleans indicant si la llista ha de contenir majúscules, minúscules, números i signes.

1. Crear llista amb conjunts de caràcters possibles en funció dels paràmetres:

        *   escollir aleatoriament una majúscula.
        *   afegir majúscules a la llista de caràcters possibles.

        *   Minúscules: chr(97) - chr(122).
        *   escollir aleatoriament una minúscula.

        *   Números: chr(48) - chr(57).
        *   escollir aleatoriament un número.
        *   afegir números a la llista de caràcters possibles.
    *   Signes especials = True:
        *   Signes especials: 33-47, 58-64, 91-96, 123-126.
        *   escollir aleatoriament un signe.
        *   afegir signes a la llista de caràcters possibles.

**Retorn:** la llista final amb tots els caràcters possibles per crear la contrassenya.

```python
def definir_llista (majus, minus, num, signes):
  llista_final = []
  llista_cars_obligatoris = []
  if majus:
    llista_cars_obligatoris += np.random.choice(caràcters_ascii(65, 90))
    llista_final += caràcters_ascii(65, 90)
  if minus:
    llista_cars_obligatoris += np.random.choice(caràcters_ascii(97, 122))
    llista_final += caràcters_ascii(97, 122)
  if num:
    llista_cars_obligatoris += np.random.choice(caràcters_ascii(48, 57))
    llista_final += caràcters_ascii(48, 57)
  if signes:
    llista_cars_obligatoris += np.random.choice(caràcters_ascii(33, 47) + caràcters_ascii(58, 64) + caràcters_ascii(91, 96) + caràcters_ascii(123, 126))
    llista_final += (caràcters_ascii(33, 47) + caràcters_ascii(58, 64) + caràcters_ascii(91, 96) + caràcters_ascii(123, 126))
  return llista_final, llista_cars_obligatoris
```

Finalment convertirem aquest diccionari a **dataframe** per poder mostrar i analitzar millor la informació.

## 🧩 **Descomponent el problema: naixement de les funcions**

### ***♻ Re-utilització de funcions:***

### ***✔ llegir_fitxer(url)***

**Paràmetre:** url del fitxer

**Passos:**

1. Llegir fitxer i guardar dades a dataframe amb les següents columnes:   
    *   equip local
    *   resultat
    *   equip visitant
2. Separar columna resultat i obtenir:
    *   gols equip local
    *   gols equip visitant
3. Esborrar columna resultat
4. Convertir gols a integer
5. Convertir dataframe a llista de diccionaris. Cada diccionari conté les dades d'un partit.

**Retorn:** la llista de partits amb les seves dades: equip local, gols local, equip visitant, gols visitant.

```python
def llegir_fitxer(url):
  #Retorna una llista de partits amb les seves dades

  # Llegim fitxer text que es troba a l'adreça url, camps separats per tabulador i  sense capçalera.
  #obtenim dataframe i assignem noms a les columnes
  columnes = ["equip_local", "resultat", "equip_visitant"]
  df = pd.read_csv(url, sep = "\t", header = None, names = columnes)

  # Separem la columna 'resultat' pel guió '-' en dues columnes noves
  gols = df['resultat'].str.split('-', expand=True)
  df['gols_local'] = gols[0]
  df['gols_visitant'] = gols[1]

  # Eliminem la columna 'resultat' original
  df.drop(columns=['resultat'], inplace=True)

  # Convertim les columnes de gols a números enters
  df['gols_local'] = df['gols_local'].astype(int)
  df['gols_visitant'] = df['gols_visitant'].astype(int)

  # Convertim cada fila del dataFrame en un diccionari on les claus són els noms de les columnes.
  llista_final = df.to_dict(orient = 'records')

  return llista_final
```

### ***✔ processar_dades(llista_inicial)***

**Paràmetre:** llista inicial amb les dades dels partits

**Passos:**
1. Inicialitzar diccionari d'equips.
2. Per cada partit :
    *   Si no existeixen crear clau al diccionari per equip local i equip visitant amb valors a 0.
    *   actualitzar gols equip local:
        *   gols a favor = gols a favor + gols local partit
        *   gols en contra = gols en contra + gols visitant partit
    *   actualitzar gols equip visitant:
        *   gols a favor = gols a favor + gols visitant partit
        *   gols en contra = gols en contra + gols local partit
    *   si gols local > gols visitant aleshores:
        *   guanyats local = guanyats local + 1
        *   perduts visitant = perduts visitant + 1
        *   punts local = punts local + 3
    *   si gols local < gols visitant aleshores:
        *   guanyats visitant = guanyats visitant + 1
        *   perduts local = perduts local + 1
        *   punts visitant = punts visitant + 3
    *   si gols local = gols visitant aleshores:
        *   empatats visitant = empatats visitant + 1
        *   empatats local = empatats local + 1
        *   punts visitant = punts visitant + 1
        *   punts local = punts local + 1

**Retorn:** el diccionari d'equips amb les seves dades: gols a favor, gols en contra, partits guanyats, partits empatats, partits perduts i punts.

```python
def processar_dades(llista_partits):
  # Retorna un diccionari d'equips amb les seves dades. Les claus del diccionari són els noms dels equips.
  # Els valors del diccionari són diccionaris amb les dades de cada equip.

  # Inicialitzem el diccionari d'equips
  dict_equips = {}
  # Recorrem la llista de partits i anem construint el diccionari d'equips
  for partit in llista_partits:
    # Obtenim nom equip local i nom equip visitant
    equip_local = partit['equip_local']
    equip_visitant = partit['equip_visitant']

    # Si no existeixen afegim al diccionari equips obtinguts amb valors inicials a 0
    if equip_local not in dict_equips:
      dict_equips[equip_local] = {
          'gols_a_favor': 0,
          'gols_en_contra': 0,
          'partits_guanyats': 0,
          'partits_empatats': 0,
          'partits_perduts': 0,
          'punts': 0
      }
    if equip_visitant not in dict_equips:
      dict_equips[equip_visitant] = {
          'gols_a_favor': 0,
          'gols_en_contra': 0,
          'partits_guanyats': 0,
          'partits_empatats': 0,
          'partits_perduts': 0,
          'punts': 0
      }

    # Actualitzem gols a favor i en contra equip local
    dict_equips[equip_local]['gols_a_favor'] += partit['gols_local']
    dict_equips[equip_local]['gols_en_contra'] += partit['gols_visitant']

    # Actualitzem gols a favor i en contra equip visitant
    dict_equips[equip_visitant]['gols_a_favor'] += partit['gols_visitant']
    dict_equips[equip_visitant]['gols_en_contra'] += partit['gols_local']

    # Actualitzem partits guanyats, empatats, perduts i punts equip local i visitant
    if partit['gols_local'] > partit['gols_visitant']:
      dict_equips[equip_local]['partits_guanyats'] += 1
      dict_equips[equip_local]['punts'] += 3
      dict_equips[equip_visitant]['partits_perduts'] += 1
    elif partit['gols_local'] < partit['gols_visitant']:
      dict_equips[equip_local]['partits_perduts'] += 1
      dict_equips[equip_visitant]['partits_guanyats'] += 1
      dict_equips[equip_visitant]['punts'] += 3
    else:
      dict_equips[equip_local]['partits_empatats'] += 1
      dict_equips[equip_local]['punts'] += 1
      dict_equips[equip_visitant]['partits_empatats'] += 1
      dict_equips[equip_visitant]['punts'] += 1

  return dict_equips
```

### ***✔ convertir_dict_a_dataframe(diccionari, índex)***

**Paràmetres:** diccionari i nom de la clau del diccionari

**Passos:**

1. crear el dataframe a partir del diccionari(índex=claus diccionari)
2. convertir índex  a columna normal

**Retorn:** el dataframe creat.

```python
def convertir_dict_a_dataframe(dict, nom_index):
  # Retorna el dataframe creat a partir del diccionari

  # Crear dataframe amb índex = claus del diccionari
  df = pd.DataFrame.from_dict(dict, orient='index')
  # Assignar nom a la columna de l'índex
  df.index.name = nom_index
  # Convertir índex en una columna i crear índex numèric
  df = df.reset_index()

  return df
```

### ***✔ analisi(dataframe)***

**Paràmetres:** dataframe amb les dades que volem analitzar

**Passos:**

Calcular i mostrar:
1. Equip més golejador: equip amb el màxim de gols a favor.
2. Equip més golejat: equip amb el màxim de gols en contra.
3. Classificació final: llista d'equpis ordenada per punts descendent



```python
def analisi(df):
  print("\n" + "="*100)
  print("ANÀLISI")
  print("="*100)
  print("Equip més golejador: " + df[df['gols_a_favor'] == df['gols_a_favor'].max()]['Equip'].to_string(index=False))
  print("Equip més golejat: " + df[df['gols_en_contra'] == df['gols_en_contra'].max()]['Equip'].to_string(index=False))
  print("Classificació final: ")
  posicio = 1
  for _, row in df.iterrows():
        print(f"   {posicio}.{row['Equip']} ")
        posicio += 1


```

```python
def main_partits ():
  llista_partits = llegir_fitxer ("https://raw.githubusercontent.com/roser740/Data-Analytics/refs/heads/main/TASQUES%20S7/historic%20partits.txt")
  equips = processar_dades(llista_partits)
  df_equips = convertir_dict_a_dataframe(equips, 'Equip')
  ordenar_dataframe(df_equips, 'punts', False)
  mostrar_dataframe (df_equips, "DATASET FINAL D'EQUIPS ORDENAT PER PUNTS")
  analisi(df_equips)
```
